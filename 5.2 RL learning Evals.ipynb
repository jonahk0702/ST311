```python
import pandas as pd
import math


df = pd.read_csv("RL_model_test_errors.csv")


print("Distance off!")
print("mean")
print(df["distance_km"].mean())
print("median")
print(df["distance_km"].median())



df['geo_score'] = 5000 * (math.e ** (-10 * df['distance_km'] / 14916.862))


print("Custom Geoguess!")
print("mean")
print(df["geo_score"].mean())
print("median")
print(df["geo_score"].median())
```

```python
import pandas as pd
import matplotlib.pyplot as plt


df['geo_score'].plot()

plt.title('Score over time')
plt.xlabel('epochs')
plt.ylabel('score')
plt.show()
```

This is not good. No real trend. If any, very slow, way too few eopchss
